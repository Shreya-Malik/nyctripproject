```python
import pandas as pd

dtype_dict = {
    'VendorID': 'float32',
    'passenger_count': 'float32',  
    'trip_distance': 'float32',
    'RatecodeID': 'float32',       
    'store_and_fwd_flag': 'category',
    'PULocationID': 'int16',
    'DOLocationID': 'int16',
    'payment_type': 'float32',     
    'fare_amount': 'float32',
    'extra': 'float32',
    'mta_tax': 'float32',
    'tip_amount': 'float32',
    'tolls_amount': 'float32',
    'improvement_surcharge': 'float32',
    'total_amount': 'float32',
    'congestion_surcharge': 'float32'
}


# Datetime columns to parse
parse_dates = ['tpep_pickup_datetime', 'tpep_dropoff_datetime']

# Load the full dataset
df = pd.read_csv(
    '../data/data.csv',
    dtype=dtype_dict,
    parse_dates=parse_dates
)

# Preview data
df.head()

# Task 1. All trips with trip_distance larger than 50
long_trips = df[df['trip_distance'] > 50]
long_trips.reset_index(drop=True, inplace=True)
print(f"Extracted {len(long_trips)} trips with distance > 50.")
long_trips
```

```python
# Task 2. Extract all trips where payment_type is missing
missing_payment = df[df['payment_type'].isna()]
missing_payment.reset_index(drop=True, inplace=True)

print(f"Found {len(missing_payment)} trips with missing payment_type.")

missing_payment
```

```python
# Task 3: Determine the number of trips for each (PULocationID, DOLocationID) pair
trip_counts = df.groupby(['PULocationID', 'DOLocationID']).size().reset_index(name='trip_count')

print(f"Found {len(trip_counts)} unique (PULocationID, DOLocationID) pairs.")

trip_counts
```

```python
# Task 4: Extract and remove rows with missing values 

# Define the relevant columns to check for missing values
check_columns = ['VendorID', 'passenger_count', 'store_and_fwd_flag', 'payment_type']

# Saving in a new dataframe bad
bad = df[df[check_columns].isna().any(axis=1)].copy()

# Remove these rows from the original dataframe
df = df.drop(index=bad.index).reset_index(drop=True)

print(f"Extracted {len(bad)} bad rows.")
print(f"Remaining rows in cleaned DataFrame: {len(df)}")
```
